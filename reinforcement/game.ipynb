```python
import cv2
import numpy as np

# Ek blank image banaye (height, width, channels)
# Yahan 400x400 ki image li hai
image = np.zeros((500, 600, 4), dtype="uint8")

# Grid settings
rows = 8
cols = 8
box_size = 50 # 400/8 = 50

for i in range(rows):
    for j in range(cols):
        # Har box ka coordinate calculate karein
        x = j * box_size
        y = i * box_size
        if (j==0&i==0):
            cv2.rectangle(image, (x, y), (x + box_size+10, y + box_size), (255, 255, 0), 1)
        # Rectangle draw karein (image, start_point, end_point, color, thickness)
        else:
            cv2.rectangle(image, (x, y), (x + box_size, y + box_size), (255, 255, 255), 1)

cv2.imshow('8x8 Grid', image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
