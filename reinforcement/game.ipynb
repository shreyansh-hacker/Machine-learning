```python
import cv2
import numpy as np

# Agent Start Position
agent_row = 7
agent_col = 0

# Goal Position
goal = (0, 7)

# Obstacles
obstacles = [
    (1, 1),
    (4, 1),
    (2, 3)
]

while True:

    # Fresh Canvas
    canvas = np.zeros((400, 400, 3), dtype=np.uint8)

    # Grid
    for i in range(0, 401, 50):
        cv2.line(canvas, (0, i), (400, i), (255, 255, 255), 1)
        cv2.line(canvas, (i, 0), (i, 400), (255, 255, 255), 1)

    # Start Cell
    cv2.rectangle(canvas, (0, 350), (50, 400), (0, 255, 255), -1)

    # Goal Cell
    cv2.rectangle(canvas, (350, 0), (400, 50), (0, 255, 0), -1)

    # Obstacles
    for row, col in obstacles:

        x1 = col * 50
        y1 = row * 50

        x2 = x1 + 50
        y2 = y1 + 50

        cv2.rectangle(
            canvas,
            (x1, y1),
            (x2, y2),
            (0, 0, 255),
            -1
        )

    # Agent
    x1 = agent_col * 50
    y1 = agent_row * 50

    x2 = x1 + 50
    y2 = y1 + 50

    cv2.rectangle(
        canvas,
        (x1, y1),
        (x2, y2),
        (255, 0, 0),
        -1
    )

    cv2.imshow("Grid World", canvas)

    key = cv2.waitKey(0)

    # Future Position
    new_row = agent_row
    new_col = agent_col

    # Movement
    if key == ord('w'):
        new_row -= 1

    elif key == ord('s'):
        new_row += 1

    elif key == ord('a'):
        new_col -= 1

    elif key == ord('d'):
        new_col += 1

    elif key == ord('q'):
        break

    # Boundary Check
    if 0 <= new_row < 8 and 0 <= new_col < 8:

        # Obstacle Check
        if (new_row, new_col) not in obstacles:

            agent_row = new_row
            agent_col = new_col

    # Goal Check
    if (agent_row, agent_col) == goal:

        print("win")

        agent_row = 7
        agent_col = 0

cv2.destroyAllWindows()
```
