```python
import cv2
import numpy as np
import random

robu_row = 7
robu_col = 0

obstacles = [
    (1, 1),
    (4, 1),
    (2, 3),
    (5,3),
    (5,4),
    (5,5)
]
starting_point=(0, 350)
ending_point =(350, 0)
action =0
Q = np.zeros((64, 4))



def reset():
    global robu_row ,robu_col
    robu_row=7
    robu_col=0
    initial_state=robu_row*8 + robu_col
    return initial_state
    
def step (action):
    global robu_col,robu_row
    if action==0 : robu_row-=1 #up 
    elif action==1 : robu_row+=1 #down
    elif action==2 : robu_col-=1 #left
    elif action==3 : robu_col+=1 #right

    next_state = robu_row*8 + robu_col

    done = False
    if (robu_row,robu_col) ==(0,7):
        reward=100
        done=True
    elif(robu_row,robu_col)in obstacles :
        reward = -100
        done=True
    else :
        reward = -1
        
    return next_state, reward ,done
    

while (robu_row, robu_col) != (0, 7):
    
    canvas = np.zeros((400,400,3),dtype=np.uint8)#BGR
    
    for i in range(0,401,50):
        cv2.line(canvas,(0,i),(400,i),(255,255,255),1)
        cv2.line(canvas,(i,0),(i,400),(255,255,255),1)
    #agent_x,agent_y=0,350
        
    cv2.rectangle(canvas,(0,350),(50,400),(0,255,255),-1)
    cv2.rectangle(canvas,(350,0),(400,50),(0,255,0),-1)
    for i in obstacles:
        x1=[i[1]][0]*50
        y1=[i[0]][0]*50
        x2=x1+50
        y2=y1+50
        cv2.rectangle(canvas,(x1,y1),(x2,y2),(0,0,255),-1)
    
    x1=robu_col*50
    y1=robu_row*50
    x2=x1+50
    y2=y1+50
    cv2.rectangle(canvas,(x1,y1),(x2,y2),(255,0,0),-1) # action performed

    # state_ = np.arange(64).reshape(8,8)[agent_y//50, agent_x//50]
    # Q_ = Q[state_][np.argmax(Q[state_])]
    
    # Q[current_state,action] = reward + gamma * Q_
    
    cv2.imshow("Q-Learning", canvas)
    cv2.waitKey(100)

    starting_point = x1,y1
    new_row=robu_row
    new_col=robu_col
    
    def valid_actions(new_row,new_col):
        valid_action=[0,1,2,3] #0 = up, 1 =down, 2 =left, 3 = right
        if new_row == 0 :
            valid_action.remove(0)
        if new_row == 7 :
            valid_action.remove(1)
        if new_col == 0:
            valid_action.remove(2)
        if new_col == 7:
            valid_action.remove(3)
        return valid_action
    
    action = np.random.choice(valid_actions(new_row,new_col))
    print(action)
    if (action==0):
        new_row=new_row-1
    if (action==1):
        new_row=new_row+1
    if (action==2):
        new_col=new_col-1
    if (action==3):
        new_col=new_col+1
    robu_row=new_row
    robu_col=new_col
    
cv2.destroyAllWindows()
```
