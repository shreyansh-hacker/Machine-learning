```python
import torch
import torch.nn as nn
```

```python
x = torch.randn(2, 3, 8, 32, 32)

print("Input Shape :", x.shape)
```

```python
conv = nn.Conv3d(

    in_channels=3,

    out_channels=16,

    kernel_size=3

)

print(conv)
```

```python
output = conv(x)

print("Output Shape :", output.shape)
```

```python
conv = nn.Conv3d(

    in_channels=3,

    out_channels=32,

    kernel_size=(3,5,5),

    stride=(1,2,2),

    padding=(1,2,2),

    bias=True

)

print(conv)
```

```python
conv = nn.Conv3d(

    3,

    8,

    kernel_size=(5,3,3)

)

output = conv(x)

print(output.shape)
```

```python
conv = nn.Conv3d(

    3,

    8,

    kernel_size=3,

    stride=2

)

output = conv(x)

print(output.shape)
```

```python
conv = nn.Conv3d(

    3,

    8,

    kernel_size=3,

    padding=1

)

output = conv(x)

print(output.shape)
```

```python
x = torch.randn(4,1,10,64,64)

conv = nn.Conv3d(

    in_channels=1,

    out_channels=32,

    kernel_size=3,

    stride=1,

    padding=1

)

y = conv(x)

print("Input :",x.shape)

print("Output:",y.shape)
```

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

x = torch.randn(1, 1, 10, 28, 28)

conv = nn.Conv3d(
    in_channels=1,
    out_channels=4,
    kernel_size=3,
    padding=1
)

y = conv(x)

print(x.shape)
print(y.shape)
```

```python
fig, ax = plt.subplots(2,5, figsize=(15,6))

# Input slices
for i in range(5):

    ax[0,i].imshow(
        x[0,0,i].detach().numpy(),
        cmap="gray"
    )

    ax[0,i].set_title(f"Input D={i}")

    ax[0,i].axis("off")

# Output slices (Channel 0)
for i in range(5):

    ax[1,i].imshow(
        y[0,0,i].detach().numpy(),
        cmap="gray"
    )

    ax[1,i].set_title(f"Output D={i}")

    ax[1,i].axis("off")

plt.tight_layout()

plt.show()
```

```python
for d in range(10):

    plt.figure(figsize=(3,3))

    plt.imshow(

        y[0,0,d].detach().numpy(),

        cmap="gray"

    )

    plt.title(f"Depth Slice {d}")

    plt.axis("off")

    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Create a 10x10x10 cube
volume = np.zeros((10,10,10))

# Put a small cube inside
volume[3:7,3:7,3:7] = 1

fig = plt.figure(figsize=(8,8))

ax = fig.add_subplot(111, projection='3d')

ax.voxels(volume, edgecolor="black")

ax.set_title("3D Input Volume")

ax.set_xlabel("Width")

ax.set_ylabel("Height")

ax.set_zlabel("Depth")

plt.show()
```

```python
kernel = np.ones((3,3,3))

fig = plt.figure(figsize=(5,5))

ax = fig.add_subplot(111, projection='3d')

ax.voxels(kernel, edgecolor="black")

ax.set_title("3×3×3 Conv3D Kernel")

plt.show()
```
