```python
import os
import pandas as pd
import numpy as np

from PIL import Image
```

```python
os.makedirs("MNIST/train", exist_ok=True)
os.makedirs("MNIST/test", exist_ok=True)

for i in range(10):

    os.makedirs(f"MNIST/train/{i}", exist_ok=True)
    os.makedirs(f"MNIST/test/{i}", exist_ok=True)

print("Folders Created Successfully")
```

```python
train_df = pd.read_csv("mnist_train.csv")

print(train_df.shape)
```

```python
for index, row in train_df.iterrows():

    label = row.iloc[0]

    image = row.iloc[1:].values.reshape(28,28).astype(np.uint8)

    img = Image.fromarray(image)

    img.save(f"MNIST/train/{label}/{index}.png")

print("Train Images Saved Successfully")
```

```python
test_df = pd.read_csv("mnist_test.csv")

print(test_df.shape)
```

```python
for index, row in test_df.iterrows():

    label = row.iloc[0]

    image = row.iloc[1:].values.reshape(28,28).astype(np.uint8)

    img = Image.fromarray(image)

    img.save(f"MNIST/test/{label}/{index}.png")

print("Test Images Saved Successfully")
```
